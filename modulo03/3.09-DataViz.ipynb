#### Q4.

Apresente um gráfico de barras com o top 10 países que mais participaram de copas do mundo, onde no eixo x devem estar o nome dos países e no eixo y a contagem de participações. Você deve separar a contagem de participações em copas femininas e masculinas, empilhando as barras de cada informação.

No exemplo de barras empilhadas da galeria do matplotlib, imagine que a parte azul são as participações do país em copas masculinas, e em laranja as participações femininas:
* https://matplotlib.org/stable/gallery/lines_bars_and_markers/bar_stacked.html

Você deve implementar essa visualização nas três bibliotecas que vimos:
* Matplotlib
* Seaborn
* Plotly

Garanta que o gráfico tenha pelo menos os atributos:
* título
* rótulos de dimensão.
* legenda

````python
# Preparação dos dados
part_1 = wc[['time_1', 'ano', 'copa']].rename(columns={'time_1': 'país'})
part_2 = wc[['time_2', 'ano', 'copa']].rename(columns={'time_2': 'país'})
todas_participacoes = pd.concat([part_1, part_2]).drop_duplicates()

contagem = todas_participacoes.groupby(['país', 'copa']).size().unstack(fill_value=0)
contagem['Total'] = contagem['Feminina'] + contagem['Masculina']
top10 = contagem.sort_values(by='Total', ascending=False).head(10).drop(columns=['Total'])
````

````python
fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(top10.index, top10['Masculina'], label='Masculina', color='royalblue')
ax.bar(top10.index, top10['Feminina'], bottom=top10['Masculina'], label='Feminina', color='darkorange')

ax.set_title('Top 10 Países com Mais Participações em Copas')
ax.set_xlabel('Países')
ax.set_ylabel('Número de Participações')
ax.legend(title='Copa')
plt.xticks(rotation=45)
plt.show()
````

````python
plt.figure(figsize=(10, 6))
# Filtrar as participações apenas para os 10 mais presentes para usar no histplot
ordem_paises = top10.index.tolist()
df_top10 = todas_participacoes[todas_participacoes['país'].isin(ordem_paises)]

sns.histplot(data=df_top10, x='país', hue='copa', multiple='stack', shrink=0.8)
plt.title('Top 10 Países com Mais Participações em Copas')
plt.xlabel('Países')
plt.ylabel('Número de Participações')
plt.xticks(rotation=45)
plt.show()
````

#### Q5.

Vamos fazer um compilado com as estatísticas históricas de copas do mundo!

Com a biblioteca de sua preferência você deve criar 4 subplots organizados em um grid de 2 linhas e 2 colunas. Eles devem conter os seguintes gráficos:
* Linha 1, coluna 1: Gráfico de barras com a quantidade de jogos que aconteceram por ano
* Linha 1, coluna 2: Gráfico de área (referências a seguir) com o total de gols por ano, separando as informações de `gols_1` e `gols_2` para distinguir gols em casa e do time visitante.
* Linha 2, coluna 1: Gráfico de área com o total de cartões por ano, separando as informações de cartões amarelos e cartões vermelhos, mas agregando cartões do time 1 ou time 2. Ou seja, uma área com `cartao_amarelo_1 + cartao_amarelo_2` e outra área com `cartao_vermelho_1 + cartao_vermelho_2`.
* Linha 2, coluna 2: Gráfico de barras com o total de gols contra por ano, somando `gols_contra_1` e `gols_contra_2`.

Referências sobre gráfico de área
* Matplotlib: https://matplotlib.org/stable/gallery/lines_bars_and_markers/stackplot_demo.html#sphx-glr-gallery-lines-bars-and-markers-stackplot-demo-py
* Pandas + Matplotlib: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.plot.area.html
* Seaborn: https://seaborn.pydata.org/generated/seaborn.objects.Area.html
* Plotly: https://plotly.com/python/filled-area-plots/

````python
import ast

# Função auxiliar para contar ocorrências dentro das strings do dataframe
def contar_itens(val):
    if pd.isna(val) or val == "": return 0
    try: return len(ast.literal_eval(str(val)))
    except: return len(str(val).split('|'))

# Extraindo os totais
wc['amarelos'] = wc['cartao_amarelo_1'].apply(contar_itens) + wc['cartao_amarelo_2'].apply(contar_itens)
wc['vermelhos'] = wc['cartao_vermelho_1'].apply(contar_itens) + wc['cartao_vermelho_2'].apply(contar_itens)
wc['contra'] = wc['gols_1_contra'].apply(contar_itens) + wc['gols_2_contra'].apply(contar_itens)

# Agrupando por ano
stats_ano = wc.groupby('ano').agg(
    jogos=('ano', 'count'),
    gols_1=('gols_1', 'sum'),
    gols_2=('gols_2', 'sum'),
    amarelos=('amarelos', 'sum'),
    vermelhos=('vermelhos', 'sum'),
    contra=('contra', 'sum')
).reset_index()

# Criando a figura e o grid 2x2
fig, axs = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('Estatísticas Históricas de Copas do Mundo', fontsize=18, weight='bold')

# [0, 0] Barras: Quantidade de Jogos
axs[0, 0].bar(stats_ano['ano'], stats_ano['jogos'], color='cornflowerblue')
axs[0, 0].set_title('Quantidade de Jogos por Ano')
axs[0, 0].set_ylabel('Nº de Jogos')

# [0, 1] Área: Total de Gols (Time 1 e Time 2)
axs[0, 1].stackplot(stats_ano['ano'], stats_ano['gols_1'], stats_ano['gols_2'], 
                    labels=['Gols Mandante (1)', 'Gols Visitante (2)'], colors=['#66c2a5', '#fc8d62'])
axs[0, 1].set_title('Total de Gols por Ano')
axs[0, 1].set_ylabel('Total de Gols')
axs[0, 1].legend(loc='upper left')

# [1, 0] Área: Cartões (Amarelos e Vermelhos)
axs[1, 0].stackplot(stats_ano['ano'], stats_ano['amarelos'], stats_ano['vermelhos'], 
                    labels=['Amarelos', 'Vermelhos'], colors=['#ffd92f', '#e5c494'])
axs[1, 0].set_title('Total de Cartões por Ano')
axs[1, 0].set_ylabel('Total de Cartões')
axs[1, 0].legend(loc='upper left')

# [1, 1] Barras: Gols Contra
axs[1, 1].bar(stats_ano['ano'], stats_ano['contra'], color='indianred')
axs[1, 1].set_title('Gols Contra por Ano')
axs[1, 1].set_ylabel('Total Gols Contra')

# Aplicando rótulo 'Ano' no eixo X apenas na linha de baixo para manter limpo
axs[1, 0].set_xlabel('Ano')
axs[1, 1].set_xlabel('Ano')

plt.tight_layout()
plt.show()
````
